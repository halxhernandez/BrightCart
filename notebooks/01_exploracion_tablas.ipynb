# 1. Carga de Datos y Perfilado Exploratorio (EDA)

En esta sección cargamos los extractos crudos de `datos/raw/` en un diccionario de DataFrames y ejecutamos una función automatizada de auditoría de datos.

```python
from pathlib import Path

import pandas as pd

# Definir rutas base del proyecto de forma dinámica
PROJECT_PATH = Path.cwd().resolve().parent
DATA_PATH = PROJECT_PATH / "datos"
RAW_DATA_PATH = DATA_PATH / "raw"
```

```python
# Diccionario que centraliza todos los datasets crudos del e-commerce

dfs = {
    "campañas":      pd.read_csv(RAW_DATA_PATH / "campana.csv"),
    "clientes":      pd.read_csv(RAW_DATA_PATH / "clientes.csv"),
    "devoluciones":  pd.read_csv(RAW_DATA_PATH / "devoluciones.csv"),
    "lineas pedido": pd.read_csv(RAW_DATA_PATH / "lineas_pedido.csv"),
    "pedidos":       pd.read_csv(RAW_DATA_PATH / "pedidos.csv"),
    "productos":     pd.read_csv(RAW_DATA_PATH / "productos.csv"),
    "visitas web":   pd.read_csv(RAW_DATA_PATH / "visitas_web.csv"),
}
```

```python
def data_profiling(df: pd.DataFrame, table_name: str):
    """
    Realiza un perfilado exploratorio rápido de un DataFrame:
    dimensiones, duplicados, tipos, nulos, valores únicos y estadísticas.
    Compatible tanto para consola como para Jupyter Notebooks.
    """

    sep = "=" * 50

    # 1. Métricas generales de la tabla
    print(sep)
    print(f"TABLA: {table_name.upper()}")
    print(sep)
    print(f"Dimensiones: {df.shape[0]:,} filas x {df.shape[1]:,} columnas")
    print(f"Filas totalmente duplicadas: {df.duplicated().sum()}")
    print(sep)

    # 2. Resumen estructurado por columna (Tipos, Nulos, % Nulos y Cardinalidad)
    print("Resumen de columnas (Tipos, Nulos y Únicos):")
    summary_df = pd.DataFrame({
        "Tipo_Dato"     : df.dtypes,
        "Nulos"         : df.isnull().sum(),
        "%_Nulos"       : ((df.isnull().sum() / len(df)) * 100).round(2),
        "Valores_Unicos": df.nunique()
    })
    print(summary_df.to_string())
    print(sep)

    # 3. Muestra visual inicial (Head)
    print("Muestra de los datos (Head):")
    try:
        from IPython.display import display
        display(df.head())
    except ImportError:
        print(df.head())

    print(sep)

    # 4. Estadísticas descriptivas globales (numéricas y categóricas)
    print("Estadísticas descriptivas:")
    try:
        from IPython.display import display
        display(df.describe(include="all"))
    except ImportError:
        print(df.describe(include="all"))
    print(sep, "\n\n")
```

```python
# Iterar sobre el diccionario para auditar cada tabla de forma automatizada
for name, df in dfs.items():
    data_profiling(df, table_name=name)
```
